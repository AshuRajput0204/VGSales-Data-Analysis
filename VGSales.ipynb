# Data Cleaning

```python
# Exact missing values in Year and Publisher columne.
df[["Year","Publisher"]].isnull().sum()
```

```python
# To check invalid year like(0,9999 or future year).
df.Year.unique()                                     # No invalid year 
```

```python
""" filtering the rows with missing year and Publisher value to determune importance if they represent only a small portion 
    of the dataset and cannot be realiably recovered,removing them is the better option."""
```

```python
df[df.Year.isna()].head(10)
```

```python
df[df.Year.isna() & df.Publisher.isna()].sample(10)
```

```python
df[df.Year.isna() & df.Publisher.isna()].shape[0]
```

```python
df=df.dropna(subset=["Publisher"])
```

```python
df.isnull().sum()
```

```python
""" 1. missing value handling      ==> 1.Year missing value      => 271 (1.63 % )
                                       2.Publisher MV            => 58  (0.35 % )
                                       3.Null in both Year & pub => 22  
    
    2. Removed rows with missing Publisher ,as they represent only 0.35 % of the dataset.
    3. Rows with missing year were retained because they still contain valid sales data.
    4. these rows will be excluded only from year-based analysis to aviod baised insights.
"""  
```
